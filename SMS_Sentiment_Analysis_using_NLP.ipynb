```python
# ===============================================================
# Project Name : SMS Sentiment Analysis using NLP
# Algorithm    : Multinomial Naive Bayes
# Author       : REEM FAYYAZ
# ===============================================================

# ===============================================================
# Step 1 : Import Required Libraries
# ===============================================================

import pandas as pd
import joblib

# Used for splitting dataset

from sklearn.model_selection import train_test_split

# Used for converting text into numerical vectors

from sklearn.feature_extraction.text import TfidfVectorizer

# Machine Learning Algorithm

from sklearn.naive_bayes import MultinomialNB

# Used for evaluating model performance

from sklearn.metrics import accuracy_score, classification_report

# ===============================================================
# Step 2 : Create SMS Dataset
# ===============================================================

data = {

"SMS":[

# ===========================
# Positive SMS
# ===========================

"Congratulations you got selected for the interview",

"Thank you for your valuable support",

"Your order has been delivered successfully",

"Happy Birthday have a wonderful year",

"Payment received successfully",

"Your appointment has been confirmed",

"You have received a cashback reward",

"Best wishes for your new journey",

# ===========================
# Negative SMS
# ===========================

"Your payment has failed",

"Transaction declined please try again",

"Your account has been temporarily blocked",

"Delivery has been cancelled",

"Your request has been rejected",

"Low balance in your account",

"Your application has been denied",

"Sorry your booking was cancelled",

# ===========================
# Neutral SMS
# ===========================

"Meeting is scheduled at 10 AM",

"Please call me when you are free",

"Your OTP is 458921",

"The train will arrive at platform 2",

"Today's temperature is 30 degrees",

"Class will begin at 9 AM",

"Your parcel is out for delivery",

"Reminder your bill is due tomorrow"

],

"Sentiment":[

# Positive

"Positive",
"Positive",
"Positive",
"Positive",
"Positive",
"Positive",
"Positive",
"Positive",

# Negative

"Negative",
"Negative",
"Negative",
"Negative",
"Negative",
"Negative",
"Negative",
"Negative",

# Neutral

"Neutral",
"Neutral",
"Neutral",
"Neutral",
"Neutral",
"Neutral",
"Neutral",
"Neutral"

]

}

# ===============================================================
# Step 3 : Convert Dictionary into DataFrame
# ===============================================================

df = pd.DataFrame(data)

print("======================================")

print("SMS Sentiment Dataset")

print("======================================")

print(df)

# ===============================================================
# Step 4 : Separate Features and Target
# ===============================================================

X = df["SMS"]

# Input SMS

y = df["Sentiment"]

# Output Label

# ===============================================================
# Step 5 : Convert Text into Numerical Features
# ===============================================================

vectorizer = TfidfVectorizer(

stop_words="english",

lowercase=True

)

X_vector = vectorizer.fit_transform(X)

print("\nVocabulary Size :", len(vectorizer.get_feature_names_out()))

# ===============================================================
# Step 6 : Split Dataset
# ===============================================================

X_train, X_test, y_train, y_test = train_test_split(

X_vector,

y,

test_size=0.20,

random_state=42

)

print("\nTraining Samples :", X_train.shape[0])

print("Testing Samples :", X_test.shape[0])

# ===============================================================
# Step 7 : Create Machine Learning Model
# ===============================================================

model = MultinomialNB()

# ===============================================================
# Step 8 : Train Model
# ===============================================================

model.fit(

X_train,

y_train

)

print("\nModel Training Completed Successfully")

# ===============================================================
# Step 9 : Prediction
# ===============================================================

prediction = model.predict(

X_test

)

# ===============================================================
# Step 10 : Evaluate Model
# ===============================================================

accuracy = accuracy_score(

y_test,

prediction

)

print("\n======================================")

print("Model Accuracy")

print("======================================")

print(accuracy)

print("\nClassification Report")

print(classification_report(

y_test,

prediction

))

# ===============================================================
# Step 11 : Predict User SMS
# ===============================================================

sms = [

input("\nEnter SMS : ")

]

sms_vector = vectorizer.transform(

sms

)

result = model.predict(

sms_vector

)

print("\nPredicted Sentiment :", result[0])

# ===============================================================
# Step 12 : Save Model
# ===============================================================

joblib.dump(

model,

"model.pkl"

)

joblib.dump(

vectorizer,

"vectorizer.pkl"

)

print("\nModel Saved Successfully")

print("model.pkl")

print("vectorizer.pkl")

# ===============================================================
# Step 13 : Display Vocabulary Size
# ===============================================================

print("\nVocabulary Size :")

print(len(vectorizer.vocabulary_))

# ===============================================================
# Step 14 : Display Sample Features
# ===============================================================

print("\nSample Features :")

print(vectorizer.get_feature_names_out()[:30])

# ===============================================================
# Step 15 : Project Completed
# ===============================================================

print("\n======================================")

print("SMS Sentiment Analysis Model is Ready")

print("======================================")
```
