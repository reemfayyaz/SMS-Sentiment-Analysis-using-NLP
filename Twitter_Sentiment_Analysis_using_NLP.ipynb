```python
# ===============================================================
# Project Name : Twitter Sentiment Analysis using NLP
# Algorithm    : Multinomial Naive Bayes
# Author       : REEM FAYYAZ
# ===============================================================

# ===============================================================
# Step 1 : Import Required Libraries
# ===============================================================

import pandas as pd
import joblib

# Used for splitting dataset

from sklearn.model_selection import train_test_split

# Used for converting text into numerical vectors

from sklearn.feature_extraction.text import TfidfVectorizer

# Machine Learning Algorithm

from sklearn.naive_bayes import MultinomialNB

# Used for evaluating the model

from sklearn.metrics import accuracy_score, classification_report

# ===============================================================
# Step 2 : Create Twitter Dataset
# ===============================================================

data = {

"Tweet":[

# ==========================
# Positive Tweets
# ==========================

"I love this new phone",
"Today is a wonderful day",
"Feeling happy after getting promoted",
"The movie was fantastic",
"Excellent customer service",
"I enjoyed my vacation",
"The food was delicious",
"This laptop works perfectly",

# ==========================
# Negative Tweets
# ==========================

"I hate this application",
"The service is very bad",
"This product is terrible",
"I am disappointed",
"My order arrived damaged",
"Very poor experience",
"The phone battery is awful",
"Worst purchase ever",

# ==========================
# Neutral Tweets
# ==========================

"I am going to the office",
"It is raining outside",
"Meeting starts at 10 AM",
"I have completed my homework",
"The train arrived on time",
"Today is Monday",
"I am reading a book",
"The exam is next week"

],

"Sentiment":[

# Positive

"Positive",
"Positive",
"Positive",
"Positive",
"Positive",
"Positive",
"Positive",
"Positive",

# Negative

"Negative",
"Negative",
"Negative",
"Negative",
"Negative",
"Negative",
"Negative",
"Negative",

# Neutral

"Neutral",
"Neutral",
"Neutral",
"Neutral",
"Neutral",
"Neutral",
"Neutral",
"Neutral"

]

}

# ===============================================================
# Step 3 : Convert Dictionary into DataFrame
# ===============================================================

df = pd.DataFrame(data)

print("=========================================")

print("Twitter Sentiment Dataset")

print("=========================================")

print(df)

# ===============================================================
# Step 4 : Separate Features and Target
# ===============================================================

X = df["Tweet"]

# Input Tweets

y = df["Sentiment"]

# Output Labels

# ===============================================================
# Step 5 : Convert Text into Numerical Features
# ===============================================================

vectorizer = TfidfVectorizer(

stop_words="english",

lowercase=True

)

X_vector = vectorizer.fit_transform(X)

print("\nVocabulary Size :",len(vectorizer.get_feature_names_out()))

# ===============================================================
# Step 6 : Split Dataset
# ===============================================================

X_train,X_test,y_train,y_test = train_test_split(

X_vector,

y,

test_size=0.20,

random_state=42

)

print("\nTraining Samples :",X_train.shape[0])

print("Testing Samples :",X_test.shape[0])

# ===============================================================
# Step 7 : Create Machine Learning Model
# ===============================================================

model = MultinomialNB()

# ===============================================================
# Step 8 : Train Model
# ===============================================================

model.fit(

X_train,

y_train

)

print("\nModel Training Completed Successfully")

# ===============================================================
# Step 9 : Prediction
# ===============================================================

prediction = model.predict(

X_test

)

# ===============================================================
# Step 10 : Evaluate Model
# ===============================================================

accuracy = accuracy_score(

y_test,

prediction

)

print("\n=========================================")

print("Model Accuracy")

print("=========================================")

print(accuracy)

print("\nClassification Report")

print(classification_report(

y_test,

prediction

))

# ===============================================================
# Step 11 : Predict User Tweet
# ===============================================================

tweet = [

input("\nEnter Tweet : ")

]

tweet_vector = vectorizer.transform(

tweet

)

result = model.predict(

tweet_vector

)

print("\nPredicted Sentiment :",result[0])

# ===============================================================
# Step 12 : Save Model
# ===============================================================

joblib.dump(

model,

"model.pkl"

)

joblib.dump(

vectorizer,

"vectorizer.pkl"

)

print("\nModel Saved Successfully")

print("model.pkl")

print("vectorizer.pkl")

# ===============================================================
# Step 13 : Display Vocabulary Size
# ===============================================================

print("\nVocabulary Size :")

print(len(vectorizer.vocabulary_))

# ===============================================================
# Step 14 : Display Sample Features
# ===============================================================

print("\nSample Features :")

print(vectorizer.get_feature_names_out()[:30])

# ===============================================================
# Step 15 : Project Completed
# ===============================================================

print("\n=========================================")

print("Twitter Sentiment Analysis Model is Ready")

print("=========================================")
```
